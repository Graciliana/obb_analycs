# Projeto OBB Analytics

## Coleta do HTML

### Objetivo

Este notebook documenta a obtenção da página HTML da Olimpíada Brasileira de Biologia (OBB).

Foi identificado que o portal utiliza mecanismos de proteção contra requisições automatizadas (HTTP 403 para requests, curl e wget), permitindo acesso apenas por navegadores.

Por esse motivo, a coleta do HTML será realizada utilizando o navegador, preservando exatamente o conteúdo renderizado que será utilizado nas próximas etapas do pipeline.

## Diagnóstico realizado

Foram realizados os seguintes testes:

- `requests → HTTP 403`
- `curl → HTTP 403`
- `wget → HTTP 403`
- `navegador → acesso permitido`

Foi observado que os links para as provas estão presentes no HTML renderizado.

Assim, optou-se por utilizar o HTML exportado pelo navegador como fonte de dados.

## Estrutura

data/

└── raw/

    └── html/

```python
# @title Criar as pastas
from pathlib import Path

HTML_DIR = Path("../data/raw/html")

HTML_DIR.mkdir(
    parents=True,
    exist_ok=True
)

print(HTML_DIR.resolve())
```

## Como obter o HTML

1. Abrir a página:

https://olimpiadasdebiologia.butantan.gov.br/provas-gabaritos-e-classificacoes

2. Pressionar CTRL + U

(Exibir código-fonte da página)

3. Salvar o arquivo

CTRL + S

Nome:

provas_obb.html

Local:

data/raw/html/

```python
# @title Verificar se exite o arquivo
arquivo_html ="/content/drive/MyDrive/Colab Notebooks/OBB_GRA/Data/raw/html/provas_obb.html"
arquivo_html = Path(arquivo_html)

print("Existe :", arquivo_html.exists())
```

```python
# @title Informação doo arquivo
print("Arquivo :", arquivo_html.name)

print("Existe :", arquivo_html.exists())

print("Tamanho:", arquivo_html.stat().st_size, "bytes")
```

```python
# @title Ler o arquivo html
with open(
    arquivo_html,
    encoding="utf-8"
) as f:

    html = f.read()

len(html)
```

```python
# @title verificar as primeiras linhas do arquivo
print(html[:1000])
```
